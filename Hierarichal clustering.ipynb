```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn import datasets
data=datasets.load_iris()
X=data.data
y=data.target
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_scale=scaler.fit_transform(X)
```

```python
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
pca_scale=pca.fit_transform(X_scale)
```

```python
plt.scatter(pca_scale[:,0],pca_scale[:,1],c=y)
```

```python
import scipy.cluster.hierarchy as sc
plt.figure(figsize=(12,8))
plt.title('dendogram')
sc.dendrogram(sc.linkage(pca_scale,method='ward'))
plt.xlabel("sample data")
plt.ylabel("dendrogram")
```

```python
from sklearn.cluster import AgglomerativeClustering
cluster=AgglomerativeClustering(n_clusters=2,linkage='ward')
cluster.fit(pca_scale)
```

```python
cluster.labels_
```

```python
from sklearn.metrics import silhouette_score
silhouette_coeff=[]
for k in range(2,11):
    cluster=AgglomerativeClustering(n_clusters=k)
    cluster.fit(pca_scale)
    score=silhouette_score(pca_scale,cluster.labels_)
    silhouette_coeff.append(score)
```
